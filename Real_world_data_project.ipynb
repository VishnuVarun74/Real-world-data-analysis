```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("retail_sales.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
print(df.columns)
```

```python
plt.figure(figsize=(10,5))

df.iloc[:, 0].value_counts().head(10).plot(kind='bar')

plt.title("Top Categories")
plt.show()
```

```python
plt.figure(figsize=(12,6))

top_values = df.iloc[:, 1].value_counts().head(10)

sns.barplot(x=top_values.index, y=top_values.values)

plt.xticks(rotation=45)
plt.title("Top 10 Categories")
plt.xlabel("Category")
plt.ylabel("Count")

plt.tight_layout()
plt.show()
```
